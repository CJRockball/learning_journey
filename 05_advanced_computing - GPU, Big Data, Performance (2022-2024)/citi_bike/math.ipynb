```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
#y=Ca^x
x = np.linspace(0,10, 20)
C1 = 1.
a1 = 2.
y1 = C1*a1**x

C2 = 1.
a2 = 1.
y2 = C2*a2**x

C3 = 1.
a3 = 1/2.
y3 = C3*a3**x
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.plot(x,y1, label='y=1*2^x')
plt.title('C=1, a=2')
plt.legend()
plt.grid()

plt.subplot(1,3,2)
plt.plot(x,y2, label='y=1*1^x')
plt.title('C=1, a=1')
plt.legend()
plt.grid()

plt.subplot(1,3,3)
plt.plot(x,y3, label='y=1*1/2^x')
plt.title('C=1, a=1/2')
plt.legend()
plt.grid()

plt.tight_layout
plt.show()
```
