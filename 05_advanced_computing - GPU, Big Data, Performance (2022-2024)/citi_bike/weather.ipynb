```python
ghcn = pd.read_csv('data/weather/weather2019.csv',
   usecols=["DATE","AWND","PRCP","SNOW","SNWD","TMAX","TMIN",
       "WT01","WT02","WT04","WT08"])

df2 = pd.read_csv('data/weather/weather2019.csv')
```

```python
display(ghcn.columns)
display(df2.columns)
```

```python
display(ghcn.head(5))
ghcn = ghcn.fillna(0)
ghcn['TMAX'] = round((ghcn['TMAX'] - 32)/1.8,1)
ghcn['TMIN'] = round((ghcn['TMIN'] - 32)/1.8,1)
col_dict = {'WT01':'fog', 'WT02':'heavy_fog', 'WT04':'sleet', 'WT08':'haze'}
ghcn.rename(columns=col_dict, inplace=True)
display(ghcn.head(5))
```

```python
int_col = ['SNOW', 'fog','heavy_fog','sleet','haze']
float_col = ['AWND','PRCP','SNWD','TMAX','TMIN']
date_col = ['DATE']

for col in float_col:
    print(f'col: {col}, min: {ghcn[col].min()}, max: {ghcn[col].max()}')
    
print(ghcn.dtypes)
```

```python
def col_type_mod(df_mod):
    for col in int_col:
        df_mod[col] = pd.to_numeric(df_mod[col], downcast='integer')

    for col in float_col:
        df_mod[col] = pd.to_numeric(df_mod[col], downcast='float')
        
    for col in date_col:
        df_mod[col] = pd.to_datetime(df_mod[col])

    return df_mod

ghcn = col_type_mod(ghcn)
print(ghcn.dtypes)
display(ghcn.head(5))
```

```python
ghcn['fsh'] = ghcn['fog'] | ghcn['heavy_fog'] | ghcn['sleet'] | ghcn['haze']
display(ghcn.head(10))
```

```python
ghcn.to_csv('data/weather/2019_clean.csv')
```
